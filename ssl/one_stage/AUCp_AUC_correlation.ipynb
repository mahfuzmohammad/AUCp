RSNA correlation and P values

```python
import pandas as pd
from scipy.stats import pearsonr
#read from csv
df = pd.read_csv('/scratch/frafsani/AUCp_experiments/MedIAnomaly/ssl/one_stage/output/rsna/fold_0/cut_paste/rsna_aucp.csv')

# find pearson correlation between aucp and auroc
# correlation = df['aucp'].corr(df['auc'], method='pearson')
# print(f'Pearson correlation between aucp and auc: {correlation}')

corr = pearsonr(df['aucp'], df['auc'])
print(f'Pearson correlation between aucp and auc: {corr[0]}  p-value: {corr[1]}')
```

```python
df= pd.read_csv('/scratch/frafsani/AUCp_experiments/MedIAnomaly/ssl/one_stage/output/rsna/fold_0/fpi/rsna_aucp.csv')
# count the number of rows in df where model path contains 'fpi_poison'
count = df['model_path'].str.contains('fpi_poisson').sum()
# take these rows and find the correlation between aucp and auc
df = df[df['model_path'].str.contains('fpi_poisson')]
print(df.head())
# find pearson correlation between aucp and auroc
corr = pearsonr(df['aucp'], df['auc'])
print(f'Pearson correlation between aucp and auc fpi_poisson: {corr[0]}  p-value: {corr[1]}')
```
